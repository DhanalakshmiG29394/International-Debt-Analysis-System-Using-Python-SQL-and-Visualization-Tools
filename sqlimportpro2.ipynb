```python
#Calculate total debt for each indicator.

query14="""
select ac.series_code, ids.indicator_name, sum(ac.debt_amt) as total_debt_value 
from all_country as ac inner join ids_data as ids on ac.series_code=ids.series_code group by ac.series_code, ids.indicator_name
order by total_debt_value desc"""

df=pd.read_sql(query14, conn)
print(df)
```

```python
#Find the country with the lowest total debt.

query16="""
select country_code, country_name, min(debt_amt) as lowest_debt from all_country group by country_code, country_name order by lowest_debt desc limit 1"""

df=pd.read_sql(query16, conn)

print(df)
```

```python
#Calculate total debt for each country and indicator combination.

query17= """
select ac.series_code, ac.country_name, ids.indicator_name, sum(ac.debt_amt) as total_debt from all_country as ac inner join ids_data as ids
on ac.series_code=ids.series_code group by ac.series_code, ac.country_name, ids.indicator_name order by total_debt asc"""

df= pd.read_sql(query17, conn)

print(df)
```

```python
#Count how many indicators each country has.

query18="""
select ac.series_code, ac.country_name, count(ids.indicator_name) as indicators from all_country as ac inner join ids_data as ids
on ac.series_code = ids.series_code group by ac.series_code, ac.country_name order by indicators desc
"""

df=pd.read_sql(query18, conn)
print(df)
```

```python
#Display countries whose total debt is above the global average.

query19 = """
select country_code, country_name, sum(debt_amt) as debt from all_country group by country_code, country_name having sum(debt_amt) > (select 
avg(debt_amt) from all_country) order by debt desc
"""

df=pd.read_sql(query19, conn)
print(df)
print(df.shape)
```

```python
#Rank countries based on total debt (highest to lowest).

query20= """
SELECT 
    country_name,
    country_code,
    SUM(debt_amt) AS total_debt,
    RANK() OVER (
        ORDER BY SUM(debt_amt) DESC
    ) AS debt_rank
FROM all_country
GROUP BY country_name, country_code
ORDER BY debt_rank"""

df=pd.read_sql(query20, conn)
print(df)


```

```python
#Find the top 5 indicators contributing most to global debt.

query21="""
select ac.series_code, sum(ac.debt_amt) as maximum_debt, ids.indicator_name
from all_country as ac inner join ids_data as ids on ac.series_code=ids.series_code
group by ac.series_code, ids.indicator_name
order by maximum_debt desc limit 5 """

cursor.execute(query21)
result=cursor.fetchall()
print(result)
```

```python
#Calculate percentage contribution of each country to total global debt.

query22= """
select country_name, sum(debt_amt) as total_debt, round(
(sum(debt_amt)/(select sum(debt_amt) from all_country)) * 100,2) as debt_percent
from all_country group by country_name order by debt_percent desc;"""

df=pd.read_sql(query22, conn)
print(df)
```

```python
#Identify the top 3 countries for each indicator based on debt.

query23 = """select ac.country_name, ac.series_code, ids.indicator_name, 
sum(ac.debt_amt) as total_debt from all_country as ac inner join ids_data as ids
on ac.series_code=ids.series_code group by ac.series_code,
ac.country_name, ids.indicator_name
order by total_debt desc limit 3"""

df=pd.read_sql(query23, conn)

print(df)
```

```python
#Find the difference between maximum and minimum debt for each country.

query24="""
select country_name, country_code, min(debt_amt) as min_debt, max(debt_amt) as max_debt, max(debt_amt) - min(debt_amt) as differnce 
from all_country group by country_code, country_name order by differnce desc"""

df=pd.read_sql(query24, conn)
print(df)
```

25 #Create a view for the top 10 countries with highest debt.
Categorize countries into:
High Debt
Medium Debt
Low Debt (based on thresholds)

```python
query25 = """
CREATE VIEW IF NOT EXISTS country_debt_cat AS

SELECT
    country_name,
    country_code,
    SUM(debt_amt) AS total_debt,

    CASE
        WHEN SUM(debt_amt) >= 4.10000 THEN 'High Debt'
        WHEN SUM(debt_amt) >= 1.87931 THEN 'Medium Debt'
        ELSE 'Low Debt'
    END AS debt_category

FROM all_country
GROUP BY country_name, country_code
"""

cursor.execute(query25)
conn.commit()

print("View created successfully")
```

```python
query25a="""select * from country_debt_cat"""

df=pd.read_sql(query25a, conn)
print(df)
```

```python
#Use window functions to calculate cumulative debt per country.

query26 = """
SELECT
    country_name,
    country_code,
    series_code,
    year,
    debt_amt,
    SUM(debt_amt) OVER (
        PARTITION BY country_code, series_code
        ORDER BY year
    ) AS cumulative_debt
FROM all_country
ORDER BY country_code, series_code, year
"""

df = pd.read_sql(query26, conn)

print(df)
```
